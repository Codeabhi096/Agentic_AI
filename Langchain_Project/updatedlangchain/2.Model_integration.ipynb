```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# Load environment variables
load_dotenv()


# Initialize Gemini model
model = init_chat_model(
    "gemini-1.5-flashs",
    temperature=0
)

# Invoke model
response = model.invoke("Tell me about the machine learning")

print(response.content)
```

batch and Stream : 

The batch() method is used when you have multiple inputs (e.g., a list of questions or prompts) and you want to process them all in parallel instead of running them one by one in a loop.

The stream() method is used when you want the model's output to appear word-by-word (in real-time chunks) rather than waiting for the entire response to generate all at once.


```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
model.stream("tell me about the artificial intelligence")
```

```python
for chunk in model.stream("Tell me aboout the Artificial super intelligence"):
    print(chunk.text)
```
